# mathmatical operation on array

```python
import numpy as np
```

```python
a=np.array([10,20,30,40])
b=np.array([50,60,70,80])
print("valuue of a",a)
print("value of b",b)
```

```python
#add():
np.add(a,b)
```

```python
# subtract(): 
np.subtract(a,b)
```

```python
# multiply():
np.multiply(a,b)
```

```python
# divide():
np.divide(a,b)
```

```python
# mod():# preform module 
# give the reminder after the division
np.mod(a,b)
```

```python
# square()
np.square(a)
np.square(b)
```

```python
# floor_divide()
np.floor_divide(a,b)
```
