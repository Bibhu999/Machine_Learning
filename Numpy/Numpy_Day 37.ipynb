```python
# standard deviation 
# stap 1: find mean
# (10 + 20 + 30 + 40)/4
# step2: find how far each value from the mean 
# value-mean
# 10-25=15
# 20-25=5
# 30-25=5
# 40-25=15
# # step3: square the difference
# (15**2+5**2+5**2+15**2)/4=125
# root of 125 = 11.1803
```

### 2. Standard deviation

* **`np.average(a)` & `np.mean(s)`**
  Calculates the arithmetic mean (average) of the values:
  $$\text{Mean} = \frac{10 + 20 + 30 + 40}{4} = 25.0$$
  *(Note: `np.average` also supports weighted averages if a `weights` parameter is provided, whereas `np.mean` does not).*

* **`np.median(a)`**
  Finds the middle value of the dataset. Since `a` has an even number of elements, it computes the average of the two middle numbers ($20$ and $30$), returning `25.0`.

* **`np.std(a)`**
  Computes the population standard deviation, measuring how spread out the values are around the mean ($25.0$):
  $$\text{Variance} = \frac{(10-25)^2 + (20-25)^2 + (30-25)^2 + (40-25)^2}{4} = \frac{225 + 25 + 25 + 225}{4} = 125$$
  $$\text{Standard Deviation} = \sqrt{125} \approx 11.1803$$
