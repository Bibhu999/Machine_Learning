# matplotlib: data visualition tool

<pre>1.matplotlib is a python library use to create different charts for visualizaing data.</pre>

```python
import matplotlib.pyplot as plt
```

# line charts:

<pre>A line chart is a graph that connected points to show how data  changes or move overtime</pre>

```python
x=[1,2,5,7,9,11,13]
y=[2,3,6,8,10,12,14]
plt.plot(x,y,c='green')
plt.title('Line chart')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.show()
```

```python
x=[1,2,5,7,9,11,13]
y=[2,3,6,8,10,12,14]
plt.plot(x,y,c='green',marker='o')
plt.title('Line chart')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.show()
```

```python
x=[1,2,5,7,9,11,13]
y=[2,3,6,8,10,12,14]
plt.plot(x,y,c='green',marker='|')
plt.title('Line chart')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.show()
```

```python
x=[1,2,5,7,9,11,13]
y=[2,3,6,8,10,12,14]
plt.plot(x,y,c='black',marker='P',linestyle='--')
plt.grid()# square boxes
plt.title('Line chart')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.show()
```

```python
# Barchart
x=[1,2,5,7,9,11,13]
y=[2,3,6,8,10,12,14]
plt.plot(x,y,c='black',marker='P',linestyle='--')
plt.grid()
plt.title('Line chart')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.show()
```

```python
## Barchart
it represents data using rectangle bar wheree the height of each bar shows the values of each category.
```

```python
x=['Team A','Team B','Team C','Team D']
y=[500,450,670,250]
plt.bar(x,y,color=['red'])
plt.title('bar plot',color='blue')
plt.xlabel('X-Axis',color='violet')
plt.ylabel('Y-Axis',color='red')
plt.show()
```

```python
# multiple bar plat
x=['Team A','Team B','Team C','Team D']
y=[500,450,670,250]
plt.bar(x,y)
x=['Team A','Team B','Team C','Team D']
y2=[550,480,700,300]
plt.bar(x,y2,width=0.7)
plt.title('bar plot',color='blue')
plt.xlabel('X-Axis',color='red')
plt.ylabel('Y-Axis',color='red')
plt.show()
```

```python
# Histogram
# A histogram is a graph that shows the distribution of numerical data by grouping values into the range.
age =[12,20,90,25,13,18,24]
interval=[20,30,40,50,60,70,80,90]
plt.title('Histogram')
plt.hist(age,interval)
plt.show()
```

# scatter plot

<pre>scatter plot is use to show the relationship between two variables each points represents one observation</pre>

```python
study_hrs=[3,4,2,5,1]
result=[70,80,50,95,10]
plt.scatter(study_hrs,result,marker='^',s=200)
plt.grid()
plt.xlabel('study_hrs')
plt.ylabel('result')
plt.show()
```

# flower plot

```python
x=[3,4,72,11,38,57,98,54,65]
y=[2,6,4,9,7,3,10,9,7]
plt.bar(x,y,width=0.7,color='g')
plt.scatter(x,y,s=300,marker='*',color='black',linewidths=7)
plt.scatter(x,y,color='y')
```

# pie plot -> it is use to display the data as preparation of my whole day

```python
activities=['studying','food','romaing','playing','sleeping','Reels']
slices=[4,2,8,3,6,1]
col=['green','violet','red','blue','brown','pink']
plt.pie(slices,labels=activities,colors=col,autopct='%1.1f%%')# to print percentage in pie  chart
plt.show()

```

```python
activities=['studying','food','romaing','playing','sleeping','Reels']
slices=[4,2,8,3,6,1]
col=['green','violet','red','blue','brown','pink']
plt.pie(slices,labels=activities,colors=col,autopct='%1.9f%%',shadow=True)
plt.show()
```

```python
activities=['studying','food','romaing','playing','sleeping','Reels']
slices=[4,2,8,3,6,1]
col=['green','violet','red','blue','brown','pink']
plt.pie(slices,labels=activities,colors=col,autopct='%1.1f%%',shadow=True,explode=(0.,0,0.2,0,0,0))# to print percentage in pie  chart
plt.show()

```
