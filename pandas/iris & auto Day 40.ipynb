# Swarm plot():usedata points which is adjusted along the categorical axis to prevent overlap every indivisual observation is visiable

```python
sns.swarmplot(x=iris.Flower_type,
             y=iris.petal_width)
plt.grid()
plt.show()
```

```python
sns.swarmplot(data=iris)
plt.grid()
plt.show()
```
