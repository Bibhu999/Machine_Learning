```python
ip=iris.drop('Flower_type',axis=1)
ip
```

```python
op=iris.Flower_type
op
```

# Train Test Split

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(ip,op,train_size=0.8)
```

```python
print(x_train.shape)
print(x_test.shape)       
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train=sc.fit_transform(x_train)# fit_transform use to fit x scale & 
x_test=sc.fit_transform(x_test)
```

```python
x_train
```

```python
x_test
```

# logistic regrassion in iris

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
lr.fit(x_train, y_train)
```

```python
# prediction 
pred=lr.predict(x_test)
pred
```

```python
from sklearn.metrics import accuracy_score, recall_score, f1_score, precision_score

acc = accuracy_score(y_test, pred)
recall = recall_score(y_test, pred, average='macro')
f1 = f1_score(y_test, pred, average='macro')
precision = precision_score(y_test, pred, average='macro')

print("Accuracy:", acc)
print("Recall score:", recall)
print("F1 score:", f1)
print("Precision score:", precision)
```

```python
# confusion matrix
from sklearn.metrics import (
    confusion_matrix, 
    ConfusionMatrixDisplay
)
cm=confusion_matrix(y_test,pred)
print(cm)
```

```python
cmd=ConfusionMatrixDisplay(cm)
cmd.plot()
plt.show()
```

# k-Nearest neighbour (KNN ALGORITHM)

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(x_train, y_train)
```

```python
pred2=knn.predict(x_test)
pred2
```

```python
from sklearn.metrics import accuracy_score, recall_score, f1_score, precision_score

acc = accuracy_score(y_test, pred2)
recall = recall_score(y_test, pred2, average='macro')
f1 = f1_score(y_test, pred2, average='macro')
precision = precision_score(y_test, pred2, average='macro')

print("Accuracy:", acc)
print("Recall score:", recall)
print("F1 score:", f1)
print("Precision score:", precision)
```

```python
# confusion matrix
from sklearn.metrics import (
    confusion_matrix, 
    ConfusionMatrixDisplay
)
ck=confusion_matrix(y_test,pred2)
print(ck)
```

```python
cmj=ConfusionMatrixDisplay(ck)
cmj.plot()
plt.show()
```
