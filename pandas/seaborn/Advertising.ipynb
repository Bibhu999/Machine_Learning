## input/output Creation:

```python
ip=Advertising.drop('Sales',axis=1)
ip
```

```python
op=Advertising.Sales
op
```

# Train Test Split

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(ip,op,train_size=0.8)
```

```python
print(x_train.shape)
print(x_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train=sc.fit_transform(x_train)# fit_transform use to fit x scale & 
x_test=sc.fit_transform(x_test)
```

```python
x_train
```

```python
x_test
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python
lr.intercept_
```

```python
lr.coef_
```

# predictions

```python
pred=lr.predict(x_test)
pred
```

```python
from sklearn.metrics import r2_score,mean_squared_error
r2=r2_score(y_test,pred)
mse=mean_squared_error(y_test,pred)
print('Accuracy:',r2)
print("mean square error:",mse)
```

```python
df = pd.DataFrame({'y_test':list(y_test),
                  'prediction':pred})
df
```

```python
# lmplot - linear model plot
sns.lmplot(x='y_test',y='prediction',
          data=df)
plt.grid()
plt.show()
```
