```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
sns.displot(wine.Ash)
plt.show()
```

```python
sns.distplot(wine.Ash)
plt.show()
```

```python
sns.histplot(wine.Ash)
plt.show()
```

```python
sns.kdeplot(wine.Ash)
plt.show
```

```python
sns.kdeplot(wine.Ash,fill=True)
plt.show
```

```python
sns.boxplot(wine.Magnesium)
plt.show
```

```python
sns.boxplot(x=wine.Ash,
           y=wine.Magnesium,palette='plasma')
plt.grid()
plt.show()
```

```python
# voilin plot
sns.violinplot(x=wine.sl_no,
           y=wine.Alcohol,palette='plasma')
plt.grid()
plt.show()
```

```python
sns.boxenplot(wine.Alcohol)
plt.grid()
plt.show()
```

```python
sns.swarmplot(x=wine.Alcohol,
             y=wine.Magnesium)
plt.grid()
plt.show()
```

```python
sns.scatterplot(x='Alcohol',
               y='Magnesium',data=wine)
plt.grid()
plt.show()
```

```python
sns.jointplot(x=wine.Alcohol,y=wine.Magnesium)
plt.grid()
plt.show()
```

```python
# count plot(): is use to count how many times each category appears in the dataset.
sns.countplot(x=wine.,data=wine)
plt.show()
```

```python
# pair plot(): a pair plot is used to see the relationship between multiple numerical column at the same time
sns.pairplot(wine,hue='sl_no')
plt.show()
```
