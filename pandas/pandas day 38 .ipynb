```python
import pandas as pd
```

```python
# to check the version of pandas 
pd.__version__
```

```python
# Series:
s=pd.Series(['a','b','c','d'])
s
```

```python
type(s)
```

```python
# Assing index position 
s=pd.Series(['a','b','c','d'],index=[10,20,30,40])
s
```

```python
type(s)
```

```python
# indexing
s[30]
```

```python
# Assigning same index number multiple time
s=pd.Series(['a','b','c','d','e'],index=[10,20,20,40,50])
s
```

```python
# slicing
s[1:5:1]
```

```python
s[20:50:1]
```

```python
# .loc->loc at the INDEX .jadi index change kariba au slicing karibaku loc use kara.
s=pd.Series(['a','b','c','d','e'],index=[10,20,20,40,50])
print(s)
s.loc[20:50:1]
```

```python
# Tuple to series
t=pd.Series((10,20,30,40))
t
t[3]
```

```python
t[-3] # series re tuple use kale indexing haba nahi
```

```python
# for -ve indexing we can use ilok:
t.iloc[-1]
```

```python
# List to series
l=pd.Series([10,20,30,40])
l
```

```python
# set to series
l=pd.Series(10,20,30,40)
l
```

```python
# Dictionary to series
d=pd.Series({'a':1,'b':2,'c':3,'d':4})
d
```

```python
d=pd.Series({'a':1,'b':2,'c':3,'d':4},index={'e','b','f','d'})
d
```

```python
# indexing
d['b']
```

```python
d[0:2:1]
```

```python
# DataFrame: multiple roes & multiple columns
d={'name':['ram','sita','hari','subha','Bhagya'],
  'Roll':[21,22,23,24,25],
  'skills': ['cooking','Travel','sleeping','study','Foodie']}
df=pd.DataFrame(d)
df
```

```python
# indexing 
df.name[1]
```

```python
df.Roll[1]
```

```python
df[0:3:1]
```

```python
df.loc[0]
```

```python
# iloc (integer location)-> use the slicing on a DataFrame
# iloc[row,column]
df.iloc[3::]
```

```python
df.iloc[::,2]
```

```python
df.iloc[::2,::] 
```

```python
df.iloc[::-1,::-1] 
```

```python
# head(): it represent top rows (by default 5) but you can change it by putting some number in () 

df.head(5) 
```

```python
df.head(3)
```

```python
# tail(): show from buttom..
df.tail(3)
```
