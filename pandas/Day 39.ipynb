# To replace all null

```python
import pandas as pd

df2 = pd.DataFrame({
    'Days': [2, np.nan, 4, 5, 6],
    'places': ['puri', 'Manali', 'Goa', 'puri', np.nan],
    'Visitora': [5500, 5000, 4890, 2500, 3400],
    'mode': ['bus', 'train', 'Bus', 'Bike', 'cycle']
})

df2
```

```python
df2.shape
```

```python
df2.info
```

```python
df2.describe
```

```python
df2.isnull()
```

```python
df2.isnull().sum()
```

```python
df2.dtypes
```

```python
df2.places.unique()
```

```python
df2.Visitora.max()
```

```python
df2.Visitora.mean()
```

```python
df2.describe()
```

```python
# fillna -> use to replace the null value
# in case of numeric column we replace 'nan' with nan/average.
df2.Days = df2.Days.fillna(8.25)
df2.describe()
```

```python
# in place of object we replace the 'nan'with top
df2.places=df2.places.fillna('puri')
df2
```

```python
df2.isnull().sum()
```

```python
df2.describe(include='all')
```
