```python
ip=pokemon.drop(['Legendary','Name','type1','type2'],axis=1)
ip
```

```python
op=pokemon['Legendary']
op
```

# Train Test Split

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(ip,op,train_size=0.8)
```

```python
print(x_train.shape)
print(x_test.shape)    
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train=sc.fit_transform(x_train)# fit_transform use to fit x scale & 
x_test=sc.transform(x_test)   
```

```python
x_train
```

```python
x_test
```

# logistic regrassion in pokemon

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
lr.fit(x_train, y_train)
```

```python
# prediction 
pred=lr.predict(x_test)
pred
```

```python
# confusion matrix
from sklearn.metrics import (
    confusion_matrix, 
    ConfusionMatrixDisplay
)
cm=confusion_matrix(y_test,pred)
print(cm)
```

```python
cmd=ConfusionMatrixDisplay(cm)
cmd.plot()
plt.show()
```
