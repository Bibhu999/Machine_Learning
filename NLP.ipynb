# pos tag (part of speech )

```python
txt1='Mount is the heighest peak in the world'
print(txt1)
```

```python
from nltk import word_tokenize
txt_token = word_tokenize(txt1)
print(txt_token)
```

```python
txt_tag=nltk.pos_tag(txt_token)
txt_tag
```

# chunking

```python
# ne_chunk_chunking (identifies places,person,organization)
```

```python
txt2='Sunder pichai is the CEO of google. statred at new York'
print(txt2)
```

```python
txt_token=word_tokenize(txt2)
print(txt_token)
```

```python
txt_tag=nltk.pos_tag(txt_token)
txt_tag
```

```python
nltk.download('maxent_ne_chunker_tab')
```

```python
nltk.download('words')
```

```python
from nltk import ne_chunk
txt_chunk=ne_chunk(txt_tag)
print(txt_chunk)
```

```python
txt3="wow !!! this is so pretty....&^%$"
txt3
```

```python
import string
for c in txt3:
    if c in string.punctuation:
        print(c,end='')
```

```python
# lambda 
punct=[c for c in txt3 if c in string.punctuation] 
print(punct)
```

```python
import string
for c in txt3:
    if c not in string.punctuation:
        print(c,end='')
```

```python
from sklearn.feature_extraction.text import CountVectorizer
txt=['I love Singing and I Love Painting']
cv=CountVectorizer()
x=cv.fit_transform(txt)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
txt=['I love Singing and I Love Painting',
    'I Love Ice_Cream']
cv=CountVectorizer()
x=cv.fit_transform(txt)
```

```python
print(x)
```

```python
cv.vocabulary_
```

```python
print(x.toarray())
```
