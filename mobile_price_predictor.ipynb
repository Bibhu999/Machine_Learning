```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(ip,op,train_size=0.8)
```

```python
from sklearn.preprocessing import StandardScaler

sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.fit_transform(x_test)
```

```python
x_train
```

```python
x_test
```

```python
from sklearn.linear_model import LinearRegression

lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python
pred=lr.predict(x_test)
pred
```

```python
from sklearn.metrics import mean_squared_error,r2_score
mse=mean_squared_error(y_test,pred)

r2=r2_score(y_test,pred)

print(mse,"mean_squared_error")
print(r2,"r2_score")
```

```python
from sklearn.preprocessing import PolynomialFeatures

# Generate polynomial and interaction features from your training/testing data
poly = PolynomialFeatures(degree=2, include_bias=False)
X_poly_train = poly.fit_transform(x_train)
X_poly_test = poly.transform(x_test)
```

```python
pred_poly=lr.predict(x_test)
pred_poly
```

```python
from sklearn.linear_model import LinearRegression

# Train a separate Linear Regression model specifically for polynomial features
lr_poly = LinearRegression()
lr_poly.fit(X_poly_train, y_train)

# Predict using the polynomial test set
pred_poly = lr_poly.predict(X_poly_test)
pred_poly
```

```python
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error

mse_poly = mean_squared_error(y_test, pred_poly)

r2_poly = r2_score(y_test, pred_poly)

print(mse_poly, "mean_squared_error")

print(r2_poly, "r2_score")
```

```python
from sklearn.linear_model import Lasso

l=Lasso()
l.fit(x_train,y_train)
```

```python
pred_ls=l.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error,r2_score,mean_absolute_error
mse=mean_squared_error(y_test,pred_ls)

r2=r2_score(y_test,pred_ls)

print(mse,"mean_squared_error")

print(r2,"r2_score")
```

```python
from sklearn.linear_model import Ridge

r=Ridge()
r.fit(x_train,y_train)
```

```python
pred_r=r.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error,r2_score,mean_absolute_error
mse=mean_squared_error(y_test,pred_r)

r2=r2_score(y_test,pred_r)

print(mse,"mean_squared_error")

print(r2,"r2_score")
```

```python
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR


scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(x_train)
X_valid_scaled = scaler.transform(x_test)


svr_model = SVR(kernel="rbf", C=1.0, epsilon=0.1)
svr_model.fit(X_train_scaled, y_train)


pred_svr = svr_model.predict(X_valid_scaled)

mse = mean_squared_error(y_test, pred_svr)
r2 = r2_score(y_test, pred_svr)

print(f"{mse:.4f} mean_squared_error")
print(f"{r2:.4f} r2_score")
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error

dt = DecisionTreeRegressor(random_state=42)
dt.fit(x_train, y_train)
```

```python
pred_dt = dt.predict(x_test)
```

```python
mse = mean_squared_error(y_test, pred_dt)

r2 = r2_score(y_test, pred_dt)

print(mse, "mean_squared_error")

print(r2, "r2_score")
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error

rf = RandomForestRegressor(n_estimators=100, random_state=42)
rf.fit(x_train, y_train)
```

```python
pred_rf = rf.predict(x_test)
```

```python
mse = mean_squared_error(y_test, pred_rf)

r2 = r2_score(y_test, pred_rf)

print(mse, "mean_squared_error")

print(r2, "r2_score")
```

```python
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error

gbr = GradientBoostingRegressor(n_estimators=100, learning_rate=0.1, random_state=42)
gbr.fit(x_train, y_train)
```

```python
pred_gbr = gbr.predict(x_test)
```

```python
mse = mean_squared_error(y_test, pred_gbr)

r2 = r2_score(y_test, pred_gbr)

print(mse, "mean_squared_error")

print(r2, "r2_score")
```
